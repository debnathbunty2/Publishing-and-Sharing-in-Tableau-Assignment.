# Publishing and Sharing in Tableau Assignment

1. What is the key difference between publishing a dashboard on Tableau Public versus Tableau Server/
Cloud in terms of accessibility and data security?

- Data Security & Privacy:
Tableau Public: All data is publicly available on the internet, and anyone can download the workbook and see the underlying data.
Tableau Server/Cloud: Data is kept private behind firewalls or within secure, authenticated sites with role-based access control (RBAC), row-level security, and Single Sign-On (SSO).

Accessibility & Sharing:
Tableau Public: Designed for public sharing; anyone with the URL can access it without a login.
Tableau Server/Cloud: Requires authorized users (licensing) to log in. Sharing is controlled via permissions within the organization.

Data Limitations:Tableau Public:
Limited to 1 million rows.Tableau Server/Cloud: No specific row limits, capable of handling large datasets and live connections.

When to Use Which:
Tableau Public: Use for personal portfolio projects, public journalism, or non-sensitive data.
Tableau Server/Cloud: Use for sensitive, proprietary company data, or when user-level permissions are required.

2. When publishing a Tableau dashboard, what happens to a live connection compared to an extract?
How does this affect data refresh and performance?

- Publishing a Tableau dashboard with a live connection maintains a direct, real-time link to the source database, updating immediately upon interaction. An extract creates a compressed, static snapshot of the data, requiring scheduled refreshes to update. Live connections favor real-time data needs but can be slower, while extracts offer superior speed and reduce database load.

Data Refresh:
Live: Data refreshes instantly as you interact with the dashboard, reflecting all updates in the source immediately.
Extract: Data is static until a refresh task is scheduled and executed on Tableau Server/Cloud to update the snapshot.

Performance:
Live: Performance is directly tied to the speed of the underlying database. Heavy queries can lead to slow load times and put a strain on the source system.
Extract: Generally provides much faster, more consistent dashboard performance because it acts as a high-speed, local data engine, reducing the burden on the backend database

3. In Tableau Server or Cloud (conceptually), which feature allows administrators to control which users
can view, interact with, or edit a published dashboard?

- In Tableau Server or Tableau Cloud, the feature that allows administrators to control which users can view, interact with, or edit a published dashboard is Permissions.

Permissions are configured as rules that determine specific capabilities—such as View, Save (Web Edit), and Download—for individual users or groups on a specific piece of content, such as a workbook, data source, or project.

How it works (conceptually):
Grant/Deny: Rules are set to Allowed, Denied, or Unspecified.
Explicit Deny: A denied permission always overrides an allowed permission.Inheritance: Content can inherit permissions from the Project it is published into, or have permissions locked to that project.

4. If you want users to view a dashboard but not download the underlying data, which permission setting
would you adjust in Tableau Server?

- Steps to Restrict Data Download:
Navigate to the project, workbook, or specific dashboard in Tableau Server.
Click Actions (...) > Permissions.Add a group or user, or select an existing rule.Set Download Summary Data and Download Data/View Data to Deny.
Click Save.

Note that the Viewer site role cannot download raw data, but still set specific capabilities to explicitly deny access to sensitive information.

6. List two best practices you should follow when preparing a dashboard for public sharing to ensure data
security and privacy.

- When preparing a dashboard for public sharing, ensuring data security and privacy is critical to prevent the exposure of sensitive information. Here are two key best practices:
Aggregate and Anonymize Data: Avoid displaying granular, row-level data that can be traced back to individuals. Instead, use aggregation techniques (e.g., showing total city sales rather than individual transactions) to hide personal details. If individual data points are necessary, anonymize them by removing or masking Personally Identifiable Information (PII) like names, email addresses, or phone numbers.
Implement Data Minimization: Only include the minimum amount of data required to meet the dashboard's purpose. Before publishing, review all data sources to ensure no sensitive fields (e.g., financial data, exact locations, internal IDs) are included, reducing the risk associated with potential data leaks.

7. After publishing a dashboard, users report a “Data not found” error.
 What is the most common cause related to data source settings?

- 1. Missing or Expired Credentials
After a dashboard is published to a server (like Power BI Service or Tableau Server), it requires its own authenticated connection to the data.Authentication Mismatch: The credentials used in the desktop environment may not have been correctly passed to or configured on the server.
Expired Tokens: For cloud data sources using OAuth, the connection token may have expired or may no longer be authorized

2. Path or Connection Detail Changes
If the published dashboard points to a specific file path or server name that has changed since the initial setup, the connection will fail.
Renamed Files: If a source file (e.g., an Excel or CSV file) was renamed or moved, the dashboard cannot locate it.
Server Updates: Changes to a database server's name or IP address will break the existing connection.

3.Permissions and Privacy LevelsUsers
viewing the dashboard must have the necessary permissions to access the underlying data source.
Access Rights: If the user viewing the report does not have permission to read the data from the source (e.g., SQL database, SharePoint), they will encounter the error.
Privacy Settings: Incorrectly set privacy levels (e.g., "Private" vs. "Public" settings in Power BI) can prevent data from loading for external users.

8. When embedding a Tableau dashboard into a secure internal portal, what two additional security
measures should be implemented to ensure only authorized users can access the embedded content.

- Tableau Connected Apps (JWT Authentication): This creates a direct trust relationship between your web application and Tableau, using JSON Web Tokens (JWT) to securely authenticate users without requiring them to sign in again. It allows your application to programmatically tell Tableau who the user is and what content they can access, providing a single sign-on (SSO) experience.
Domain Allowlisting (Embedding Constraints): Within the Tableau Connected App settings, you must specify the exact domain(s) of your internal portal. This restriction ensures that even if someone steals the embedding code or JWT, they cannot embed the dashboard on a different, unauthorized website.

Note: For maximum data protection within the dashboard itself, Row-Level Security (RLS) (via user filters) should also be implemented to restrict what data each user can see, even if they have access to the dashboard container.
